<font size="5">3. Feature Engineering</font>


```python
# Here I kept only those rows where both budget and revenue are actually reported
df_ml = df[(df['budget'] > 0) & (df['revenue'] > 0)].copy()

print("Rows before filtering:", df.shape[0])
print("Rows after filtering:", df_ml.shape[0])
```

```python
# Finally I created the target column
df_ml['profit'] = df_ml['revenue'] - df_ml['budget']
df_ml['is_hit'] = (df_ml['revenue'] > df_ml['budget']).astype(int)

df_ml['is_hit'].value_counts(normalize=True)
```

```python
# Here I calculated some extra features like number of genres and release year
df_ml['genre_count'] = df_ml['genres'].str.split().apply(len)
df_ml['release_year'] = pd.to_datetime(df_ml['release_date'], errors='coerce').dt.year

df_ml[['title', 'budget', 'revenue', 'profit', 'is_hit', 'genre_count', 'release_year']].head()
```

```python
# The top 10 genres
top10_genres = df['genres'].str.split().explode().value_counts().head(10).index.tolist()

for genre in top10_genres:
    df_ml[f'genre_{genre}'] = df_ml['genres'].apply(lambda x: 1 if genre in x.split() else 0)

print(top10_genres)
```

```python
# Here i handled few missing runtime values for few rows
df_ml['runtime'] = df_ml['runtime'].fillna(df_ml['runtime'].median())

numeric_features = ['budget', 'popularity', 'runtime', 'vote_average', 'vote_count', 'genre_count', 'release_year']
genre_features = [f'genre_{g}' for g in top10_genres]

feature_cols = numeric_features + genre_features
target_col = 'is_hit'

df_ml[feature_cols + [target_col]].isna().sum()
```

```python
df_ml = df_ml.dropna(subset=feature_cols)
print("Final shape:", df_ml.shape)
```

```python
from sklearn.preprocessing import StandardScaler

X = df_ml[feature_cols]
y = df_ml[target_col]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(X_scaled, columns=feature_cols)
X_scaled.head()
```

<font size="5">4. Testing with multiple Supervised Learning algorithms</font>


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print("Train shape:", X_train.shape, "Test shape:", X_test.shape)
print("Train class balance:\n", y_train.value_counts(normalize=True))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix
```

```python
results = []  

# Logistic Regression model
for C_val in [0.01, 0.1, 1, 10]:
    model = LogisticRegression(C=C_val, max_iter=1000)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    f1 = f1_score(y_test, preds)
    results.append(('LogisticRegression', f'C={C_val}', acc, f1))
    print(f"C={C_val:<5} | Accuracy={acc:.3f} | F1={f1:.3f}")
```

```python
# Decision Tree model
for depth in [3, 5, 10, None]:
    model = DecisionTreeClassifier(max_depth=depth, random_state=42)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    f1 = f1_score(y_test, preds)
    results.append(('DecisionTree', f'max_depth={depth}', acc, f1))
    print(f"max_depth={str(depth):<5} | Accuracy={acc:.3f} | F1={f1:.3f}")
```

```python
# Random Forest model
for n_est in [100, 300]:
    for depth in [5, None]:
        model = RandomForestClassifier(n_estimators=n_est, max_depth=depth, random_state=42)
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        acc = accuracy_score(y_test, preds)
        f1 = f1_score(y_test, preds)
        results.append(('RandomForest', f'n={n_est}, depth={depth}', acc, f1))
        print(f"n_estimators={n_est}, max_depth={depth} | Accuracy={acc:.3f} | F1={f1:.3f}")
```

```python
# Finally here I computed the best model 
best_rf = RandomForestClassifier(n_estimators=300, max_depth=None, random_state=42)
best_rf.fit(X_train, y_train)
preds = best_rf.predict(X_test)

print(classification_report(y_test, preds))

cm = confusion_matrix(y_test, preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Flop','Hit'], yticklabels=['Flop','Hit'])
plt.title('Confusion Matrix — Random Forest with (n=300)')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

```python
results_df = pd.DataFrame(results, columns=['Model', 'Params', 'Accuracy', 'F1'])
results_df['Label'] = results_df['Model'] + ' (' + results_df['Params'] + ')'

plt.figure(figsize=(10, 8))
sns.barplot(data=results_df, x='Accuracy', y='Label', palette='mako')
plt.title('Test Accuracy across Algorithms & Parameter Settings')
plt.xlim(0, 1)
plt.show()

results_df.sort_values('Accuracy', ascending=False)
```

Random Forest gave the best accuracy and F1 among all algorithm-parameter combos tested, while simpler models like Logistic Regression and Decision Trees underfit slightly — showing that hit/flop prediction needs non-linear interactions between budget, popularity, and genre rather than a straight line boundary.

<font size="5">5. Cross Validation</font>


```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_results = []  
```

```python
# Logistic Regression CV
for C_val in [0.1, 1, 10]:
    model = LogisticRegression(C=C_val, max_iter=1000)
    scores = cross_val_score(model, X_train, y_train, cv=cv_strategy, scoring='accuracy')
    cv_results.append(('LogisticRegression', f'C={C_val}', scores))
    print(f"C={C_val:<5} | CV Accuracy: {scores.mean():.3f} ± {scores.std():.3f}")
```

```python
# Decision Tree CV
for depth in [5, 10, None]:
    model = DecisionTreeClassifier(max_depth=depth, random_state=42)
    scores = cross_val_score(model, X_train, y_train, cv=cv_strategy, scoring='accuracy')
    cv_results.append(('DecisionTree', f'max_depth={depth}', scores))
    print(f"max_depth={str(depth):<5} | CV Accuracy: {scores.mean():.3f} ± {scores.std():.3f}")
```

```python
# Random Forest CV
for n_est in [100, 300]:
    model = RandomForestClassifier(n_estimators=n_est, random_state=42)
    scores = cross_val_score(model, X_train, y_train, cv=cv_strategy, scoring='accuracy')
    cv_results.append(('RandomForest', f'n_estimators={n_est}', scores))
    print(f"n_estimators={n_est:<5} | CV Accuracy: {scores.mean():.3f} ± {scores.std():.3f}")
```

```python
# Graph to show all CV scores spread across models
cv_plot_data = []
labels = []
for name, params, scores in cv_results:
    cv_plot_data.append(scores)
    labels.append(f'{name}\n({params})')

plt.figure(figsize=(12, 6))
plt.boxplot(cv_plot_data, labels=labels)
plt.xticks(rotation=45, ha='right')
plt.ylabel('CV Accuracy')
plt.title('5-Fold Cross-Validation Accuracy Distribution per parameter')
plt.tight_layout()
plt.show()
```

<font size="5">6. Different Boosting Techniques</font>


```python
from sklearn.ensemble import AdaBoostClassifier
from xgboost import XGBClassifier
from sklearn.metrics import roc_curve, roc_auc_score
```

```python
# Adaboost
for n_est in [50, 100]:
    for lr in [0.5, 1.0]:
        model = AdaBoostClassifier(n_estimators=n_est, learning_rate=lr, random_state=42)
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        acc = accuracy_score(y_test, preds)
        f1 = f1_score(y_test, preds)
        results.append(('AdaBoost', f'n={n_est}, lr={lr}', acc, f1))
        print(f"n_estimators={n_est}, learning_rate={lr} | Accuracy={acc:.3f} | F1={f1:.3f}")
```

```python
# XGBoost
for depth in [3, 6]:
    for lr in [0.1, 0.3]:
        for n_est in [100, 300]:
            model = XGBClassifier(
                max_depth=depth, learning_rate=lr, n_estimators=n_est,
                use_label_encoder=False, eval_metric='logloss', random_state=42
            )
            model.fit(X_train, y_train)
            preds = model.predict(X_test)
            acc = accuracy_score(y_test, preds)
            f1 = f1_score(y_test, preds)
            results.append(('XGBoost', f'depth={depth}, lr={lr}, n={n_est}', acc, f1))
            print(f"depth={depth}, lr={lr}, n_estimators={n_est} | Accuracy={acc:.3f} | F1={f1:.3f}")
```

```python
# This tells us which features are important
best_xgb = XGBClassifier(max_depth=6, learning_rate=0.1, n_estimators=300,
                          use_label_encoder=False, eval_metric='logloss', random_state=42)
best_xgb.fit(X_train, y_train)

importances = pd.Series(best_xgb.feature_importances_, index=feature_cols).sort_values(ascending=False)

plt.figure(figsize=(8, 6))
sns.barplot(x=importances.values, y=importances.index, palette='rocket')
plt.title('XGBoost Feature Importance')
plt.show()
```

```python
# Here I compared RandomForest , Adaboost , XGBoost with ROC curve

models_for_roc = {
    'RandomForest': best_rf,
    'AdaBoost': AdaBoostClassifier(n_estimators=100, learning_rate=1.0, random_state=42).fit(X_train, y_train),
    'XGBoost': best_xgb
}

plt.figure(figsize=(8, 6))
for name, model in models_for_roc.items():
    probs = model.predict_proba(X_test)[:, 1]
    fpr, tpr, _ = roc_curve(y_test, probs)
    auc = roc_auc_score(y_test, probs)
    plt.plot(fpr, tpr, label=f'{name} (AUC={auc:.2f})')

plt.plot([0, 1], [0, 1], 'k--', label='Random guess')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison')
plt.legend()
plt.show()
```

```python
# comparison of all algorithms 

results_df = pd.DataFrame(results, columns=['Model', 'Params', 'Accuracy', 'F1'])
results_df.sort_values('Accuracy', ascending=False).head(10)
```

<font size="5">7. Testing with multiple Unsupervised Learning algorithms</font>


```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
inertias = []
silhouette_scores = []
k_range = range(2, 8)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(k_range, inertias, marker='o')
axes[0].set_xlabel('Number of Clusters (k)')
axes[0].set_ylabel('Inertia')
axes[0].set_title('Elbow Method')

axes[1].plot(k_range, silhouette_scores, marker='o', color='green')
axes[1].set_xlabel('Number of Clusters (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Score vs k')

plt.tight_layout()
plt.show()
```

```python
optimal_k = 4 

kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
df_ml['cluster'] = kmeans.fit_predict(X_scaled)

df_ml['cluster'].value_counts()
```

```python
cluster_profile = df_ml.groupby('cluster')[['budget', 'revenue', 'popularity', 'vote_average', 'runtime']].mean()
cluster_profile
```

```python
cluster_profile.plot(kind='bar', figsize=(12, 6), subplots=True, layout=(2, 3), legend=False)
plt.suptitle('Cluster Profiles — Mean Feature Values per Cluster')
plt.tight_layout()
plt.show()
```

<font size="5">8. Dimensionality reduction</font>


```python
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance captured:", pca.explained_variance_ratio_.sum())
```

```python
# PCA using Kmeans cluster

plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df_ml['cluster'], cmap='viridis', alpha=0.6)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA — Colored by KMeans Cluster')
plt.colorbar(scatter, label='Cluster')
plt.show()
```

```python
# PCA using is_hit label
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='coolwarm', alpha=0.6)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA — Colored by Hit/Flop Label')
plt.colorbar(scatter, label='is_hit')
plt.show()
```

```python
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X_scaled, y)

plt.figure(figsize=(8, 5))
plt.hist(X_lda[y == 0], bins=30, alpha=0.6, label='Flop', color='red')
plt.hist(X_lda[y == 1], bins=30, alpha=0.6, label='Hit', color='green')
plt.xlabel('LDA Component 1')
plt.ylabel('Count')
plt.title('LDA — Class Separation for Hit vs Flop')
plt.legend()
plt.show()
```

<font size="5">Overall summary</font>

<font size="3">The best model we have is the Random Forest model. It has 100 estimators and no maximum depth.

* Test accuracy is 81.9%. This means that out of 100 times the model guesses if a film is popular or not 82 times correctly.

* The F1 score of this model is 0.883. This score shows a balance between precision and recall. We focus mainly on the "hit" class

* The accuracy of cross-validation is 81.3% ± 2.0%. This is the same as the test accuracy. The small standard deviation shows that the models performance is stable. It is not a coincidence from the train/test split.

* The ROC-AUC score is 0.84. This score is greater than 0.5, which is a random guess. So the model is good at telling hits and flops apart for thresholds.

* The top 4 predictors are vote count, genre Thriller, release year and genre Action. This tells us that a movies success is mainly about votes and genre, not budget.

* PCA captured 34.6% of the variance with 2 components. This means our features do not have redundancy. This is why the tree-based model, Random Forest works better than models. Linear models can be reduced to 2 dimensions, by PCA.</font>
